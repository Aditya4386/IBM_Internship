```python
data.drop(columns=['fnlwgt'],axis=1,inplace=True)
data.drop(columns=['race'],axis=1,inplace=True)
data.drop(columns=['education'],axis=1,inplace=True)
data.head(3)
```

```python
for col in data.select_dtypes(include='object').columns:
    print(f"? in {col}: {(data[col] == '?').sum()} ")
```

```python
data.replace("?", np.nan, inplace=True)
```

```python
data=data[data['workclass'] != 'Never-worked']
data=data[data['workclass'] != 'Without-pay']

data.workclass.value_counts()
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
data.duplicated().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
categorical_cols = data.select_dtypes(include='object').columns.drop('income')

for col in categorical_cols:
    data[col] = le.fit_transform(data[col])

data['income'] = le.fit_transform(data['income']) 
data.head()
```

```python
plt.figure(figsize=(12,10))
sns.heatmap(data.corr(),annot=True)
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

features = data.drop('income', axis=1)
output_column = data['income']

scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

features_scaled_df = pd.DataFrame(features_scaled, columns=features.columns)
```

```python
features_scaled_df.head(3)
```

```python
selected_features = [
    'educational-num',
    'capital-gain',
    'capital-loss',
    'hours-per-week',
    'age'
]

x = features_scaled_df[selected_features]
y = output_column
```

```python
x.head(3)
```

```python
y.head(3)
```

```python
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42, stratify=y
)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
rf_model = RandomForestClassifier(random_state=42)
rf_model.fit(x_train, y_train)
```

```python

y_pred = rf_model.predict(x_test)

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print("Accuracy Score:", accuracy_score(y_test, y_pred)*100,"%")
print("\nClassification Report:\n", classification_report(y_test, y_pred
```

```python
mapping = {0: '<=50K', 1: '>50K'}

actual_income = [mapping[val] for val in y_test]
predicted_income = [mapping[val] for val in y_pred]

comparison_df = pd.DataFrame({
    'Actual Income': actual_income,
    'Predicted Income': predicted_income
})

print(comparison_df.head(10))
```

```python
cm = confusion_matrix(y_test, y_pred)

labels = ['<=50K', '>50K']

plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix Heatmap')
plt.show()
```

```python
income_mapping = {0: '<=50K', 1: '>50K'}

print("📋 Enter your details to predict the income category:")

educational_num = int(input("Educational Num (e.g., 10): "))
capital_gain = int(input("Capital Gain (e.g., 0): "))
capital_loss = int(input("Capital Loss (e.g., 0): "))
hours_per_week = int(input("Hours per Week (e.g., 40): "))
age = int(input("Age (e.g., 35): "))

user_input_df = pd.DataFrame([{
    'educational-num': educational_num,
    'capital-gain': capital_gain,
    'capital-loss': capital_loss,
    'hours-per-week': hours_per_week,
    'age': age
}])

prediction = rf_model.predict(user_input_df)[0]
predicted_income = income_mapping[prediction]

print(f"\n💰 Predicted Salary Category: {predicted_income}")
```
